```python
#Import Libraries
import pandas as pd #Handling the Dataframe 
import numpy as np # For Calculations 
```

```python
#Task1 Employee Dataframe
emp = pd.DataFrame({
    "ID": ["A001","A002","A003","A004","A005"],
    "Name": ["John Alter","Alice Luxumberg","Tom Sabestine","Nina Adgra","Amy Johny"],
    "Gender": ["M","F","M","F","F"],
    "City": ["Paris","London","Berlin","Newyork","Madrid"],
    "Age": [25,27,29,31,30]})
```

```python
#Seniorty Level Dataframe
seniority = pd.DataFrame({
    "ID": ["A001","A002","A003","A004","A005"],
    "Designation Level": [2,2,3,2,3]})
```

```python
#Project Dataframe (notice missing costs)
project = pd.DataFrame({
    "ID": ["A001","A002","A003","A004","A005","A002","A005","A003","A001","A003","A001","A004","A004","A005"],
    "Project": ["Project 1","Project 2","Project 3","Project 4","Project 5","Project 6","Project 7","Project 8","Project 9","Project 10","Project 11","Project 12","Project 13","Project 14"],
    "Cost": [1002000,2000000,4500000,5500000,np.nan,680000,400000,350000,np.nan,300000,2000000,1000000,3000000,200000],
    "Status": ["Finished","Ongoing","Finished","Ongoing","Finished","Failed","Finished","Failed","Ongoing","Finished","Failed","Ongoing","Finished","Finished"]})
```

```python
#Save CSV 
emp.to_csv("Employee.csv", index=False)
seniority.to_csv("Seniority.csv", index=False)
project.to_csv("Project.csv", index=False)
```

```python
#Task 2 Fill Missing Cost with Running Average
proj = pd.read_csv("Project.csv") # Reading Project CSV file

running_sum = 0 # Initializing running sum
count = 0 # Initializing count of valid values

for i in range(len(proj)): # Looping through each row
    if not pd.isna(proj.loc[i, "Cost"]):
        running_sum += proj.loc[i, "Cost"]
        count += 1
    else:
        proj.loc[i, "Cost"] = running_sum / count
proj.head(10)  # Replacing missing cost with running average
```

```python
#Task 3 Split Name Column 
emp = pd.read_csv("Employee.csv")

emp[["First Name","Last Name"]] = emp["Name"].str.split(" ", expand=True) # Splitting Name column into First and Last name
emp.drop("Name", axis=1, inplace=True) # Dropping original Name column
emp.head()
```

```python
#Task 4 Merge all the Dataframe
final = pd.merge(emp, seniority, on="ID") # Merging employee and seniority data
final = pd.merge(final, proj, on="ID") # Merging above result with project data
final.head()
```

```python
#Task 5 Add 5% Bonus for Finished Projects
final["Bonus"] = np.where(final["Status"]=="Finished", final["Cost"]*0.05, 0) # Creating Bonus column (5% of cost for Finished projects)
final[["First Name","Project","Status","Cost","Bonus"]].head()
```

```python
#Task 6 Demote 
final.loc[final["Status"]=="Failed","Designation Level"] += 1 # Increasing designation level by 1 for failed projects
final = final[final["Designation Level"] <= 4] # Removing employees with designation level above 4
final[["First Name","Status","Designation Level"]].head()
```

```python
#Task 7 Mr./Mrs. & Drop Gender 
final["First Name"] = np.where(final["Gender"]=="M","Mr. "+final["First Name"],"Mrs. "+final["First Name"]) # Adding Mr. or Mrs. based on gender
final.drop("Gender", axis=1, inplace=True) # Dropping Gender column
final.head()
```

```python
#Task 8 promote if age>29 
final.loc[final["Age"]>29,"Designation Level"] -= 1 # Reducing designation level by 1 for age greater than 29
final[["First Name","Age","Designation Level"]].head()
```

```python
#Task 9 Total Project Cost per Employee 
total_cost = final.groupby(["ID","First Name"])["Cost"].sum().reset_index() # Grouping by ID and First Name and summing cost
total_cost.rename(columns={"Cost":"Total Cost"}, inplace=True) # Renaming column
total_cost
```

```python
#Task 10 City Contains letter "o"
result = final[final["City"].str.contains("o", case=False)] # Filtering employees where city contains letter 'o'
result # Displaying result
```
